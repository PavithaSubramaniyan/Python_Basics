```python
#abstraction
from abc import ABC,abstractmethod
class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass
class Dog(Animal):
  def sound(self):
    print("Barking")
d = Dog()
d.sound()
```

```python
#polymorphism
class Dog():
  def sound(self):
    return "Dog Barking"
class Cat():
  def sound(self):
    return "Cat meow"
animal = [Dog(), Cat()]
for i in animal:
  print(i.sound())
```

```python
#Encapsulation
class Bank:
  def __init__(self, balance):
    self.__balance = balance
  def setamount(self, amount):
    self.__balance += amount
  def getamount(self):
    return self.__balance
b = Bank(100)
b.setamount(30)
print(b.getamount())
print(b._Bank__balance)#internal access , name mangling
```

```python
#inheritance
#Single inheritance
class call:
  def call(self):
    return "we can call through this"
class contact(call):
  def contact(self):
    return "We can contact them"
con = contact()
print(con.contact())
print(con.call())
```

```python
#multiple inheritance
class msg:
  def msg(self):
    return "We can message"
class call:
  def call(self):
    return "We can call"
class Whatsapp(call, msg):
  def whatsapp(self):
    return "We can whatsapp"
what = Whatsapp()
print(what.whatsapp())
print(what.msg())
print(what.call())
```

```python
#Hirerical inheritance
class whatsapp:
  def Whatsapp(self):
    return "We can whatsapp"
class call(whatsapp):
  def call(self):
    return "We can call"
class msg(whatsapp):
  def msg(self):
    return "We can msg"
c = call()
m = msg()
print(c.call())
print(c.Whatsapp())
print(m.msg())
print(m.Whatsapp())
```

```python
#multilevel inheritance
class Nokia:
  def nokia(self):
    return "old model"
class smartphone(Nokia):
  def smart(self):
    return "Smart phone"
class iphone(smartphone):
  def iphone(self):
    return "iphone"
iph = iphone()
print(iph.iphone())
print(iph.smart())
print(iph.nokia())
```

```python
#method overriding and super
class Appartment:
  def kitchen(self):
    return ("Can cook")
  def hall(self):
    return "Visiting area"
class bungalow(Appartment):
  def kitchen(self):
    return (f"bungalow kitchen: {super().kitchen()}")
  def hall(self):
    return f"bungalow hall: {super().hall()}"
b = bungalow()
print(b.kitchen())
print(b.hall())
```

```python
#mro = method resolution order - diamond pattern
class communicationApp:
  def comm(self):
    return "Communication app"
class messager(communicationApp):
  def comm(self):
    return "Messager app"
class insta(communicationApp):
  def comm(self):
    return "insta app"
class imo(messager, insta):#left to right
  def comm(self):
    return "imo app"
i = imo()
print(i.comm())#own method wins , override parents
print(imo.mro())#it shows search order
```

```python
#multithreading
from threading import Thread
def microwavehot(num):
  for i in range(num):
    print (i)
def microwave_boil(num):
  for i in range(num):
    print(i)
def micro_heat():
  for i in range(3):
    print("Completed",i)
mw = Thread(target = microwavehot, args = (5,) )
mb = Thread(target = microwave_boil, args=(5,))
mh = Thread(target=micro_heat)
mw.start()
mb.start()
mh.start()
mw.join()
mb.join()
mh.join()
```
